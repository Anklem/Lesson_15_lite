Эти команды покрывают основные аспекты работы с проектом Django и используются на разных этапах разработки, от создания приложения до работы с миграциями и базой данных.

---

### **1. `django-admin startproject <project_name>`**
- **Описание**: Создаёт новый проект Django с заданным именем.
- **Пример**: 
  ```shell
  django-admin startproject myproject
  ```
  Эта команда создаёт структуру директорий и файлов, необходимых для начала работы над проектом.

---

### **2. `python manage.py runserver`**
- **Описание**: Запускает встроенный сервер разработки Django.
- **Параметры**:
  - По умолчанию сервер запускается на порту `8000`. Если хотите запустить на другом порту:
    ```shell
    python manage.py runserver 8080
    ```
  - Также можно указать IP-адрес и порт:
    ```shell
    python manage.py runserver 0.0.0.0:8000
    ```
- **Пример**: 
  ```shell
  python manage.py runserver
  ```

---

### **3. `python manage.py startapp <app_name>`**
- **Описание**: Создаёт новое приложение в вашем проекте Django.
- **Пример**:
  ```shell
  python manage.py startapp blog
  ```
  Это создаст структуру папок для нового приложения, которое можно будет подключить к основному проекту через `settings.py`.

---

### **4. `python manage.py migrate`**
- **Описание**: Применяет все миграции базы данных, создавая или изменяя таблицы в соответствии с вашими моделями.
- **Пример**:
  ```shell
  python manage.py migrate
  ```
  Эта команда будет выполнять все доступные миграции и синхронизирует базу данных с вашим кодом моделей.

---

### **5. `python manage.py makemigrations`**
- **Описание**: Создаёт новые файлы миграций на основе изменений в моделях.
- **Пример**:
  ```shell
  python manage.py makemigrations
  ```
  После добавления или изменения моделей эта команда создаёт файл миграции, который затем можно применить с помощью команды `migrate`.

---

### **6. `python manage.py createsuperuser`**
- **Описание**: Создаёт суперпользователя для доступа к административной панели Django.
- **Пример**:
  ```shell
  python manage.py createsuperuser
  ```
  При выполнении команды система предложит ввести имя пользователя, электронную почту и пароль.

---

### **7. `python manage.py shell`**
- **Описание**: Открывает интерактивную оболочку Python с загрузкой окружения Django. Это полезно для быстрого тестирования моделей и взаимодействия с базой данных.
- **Пример**:
  ```shell
  python manage.py shell
  ```

---

### **8. `python manage.py showmigrations`**
- **Описание**: Показывает список миграций и их статус (применены ли они).
- **Пример**:
  ```shell
  python manage.py showmigrations
  ```

---

### **9. `python manage.py sqlmigrate <app_name> <migration_number>`**
- **Описание**: Показывает SQL-код, который будет выполнен для конкретной миграции.
- **Пример**:
  ```shell
  python manage.py sqlmigrate blog 0001
  ```
  Эта команда покажет SQL-запросы, которые будут выполнены для миграции `0001` в приложении `views.py`.

---

### **10. `python manage.py flush`**
- **Описание**: Очищает все данные из базы данных, но не удаляет таблицы. Используется для сброса базы данных.
- **Пример**:
  ```shell
  python manage.py flush
  ```

---

### **11. `python manage.py dbshell`**
- **Описание**: Открывает интерактивную оболочку базы данных, позволяя напрямую выполнять SQL-запросы.
- **Пример**:
  ```shell
  python manage.py dbshell
  ```

---

### **12. `python manage.py check`**
- **Описание**: Выполняет проверку проекта на наличие ошибок и предупреждений. Полезно для проверки структуры проекта перед развертыванием.
- **Пример**:
  ```shell
  python manage.py check
  ```

---

### **13. `python manage.py test`**
- **Описание**: Запускает тесты для вашего проекта. Django включает систему тестирования для проверки функциональности приложения.
- **Пример**:
  ```shell
  python manage.py test
  ```

---

### **14. `python manage.py collectstatic`**
- **Описание**: Собирает все статические файлы проекта (CSS, JS, изображения) и копирует их в одну директорию, определённую в настройках (`STATIC_ROOT`).
- **Пример**:
  ```shell
  python manage.py collectstatic
  ```

---

### **15. `python manage.py diffsettings`**
- **Описание**: Показывает различия между настройками по умолчанию Django и текущими настройками проекта.
- **Пример**:
  ```shell
  python manage.py diffsettings
  ```

---

### **16. `python manage.py dumpdata <app_name> [model]`**
- **Описание**: Экспортирует данные из указанной модели или всего приложения в формат JSON, XML или YAML.
- **Пример**:
  ```shell
  python manage.py dumpdata blog
  ```

---

### **17. `python manage.py loaddata <filename>`**
- **Описание**: Загружает данные из файла в формате JSON, XML или YAML в базу данных. Используется для восстановления данных, созданных командой `dumpdata`.
- **Пример**:
  ```shell
  python manage.py loaddata blog_data.json
  ```

---

### **18. `python manage.py dbshell`**
- **Описание**: Открывает оболочку для работы с базой данных напрямую через SQL.
- **Пример**:
  ```shell
  python manage.py dbshell
  ```

---

### **19. `python manage.py clear_cache`**
- **Описание**: Очищает кеш приложения.
- **Пример**:
  ```shell
  python manage.py clear_cache
  ```

---

### **20. `python manage.py changepassword <username>`**
- **Описание**: Меняет пароль пользователя, указав его имя.
- **Пример**:
  ```shell
  python manage.py changepassword admin
  ```
