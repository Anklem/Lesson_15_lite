## **2. Установка и настройка окружения**

### **2.1 Предварительные требования**

Перед тем как начать работу с Django, необходимо убедиться, что на вашей машине установлены необходимые инструменты и пакеты. В этой секции рассмотрим, какие компоненты должны быть установлены и почему важно использовать виртуальные окружения при разработке на Python.

#### **1. Установка Python**

Django работает на Python, поэтому важно, чтобы Python был установлен на вашем компьютере. На момент написания лекции, Django поддерживает версии Python начиная с 3.7 и выше. Чтобы проверить, установлен ли Python, откройте терминал и выполните команду:

```shell
python --version
```

Если Python установлен, вы увидите номер версии. В противном случае нужно скачать и установить Python с официального сайта [python.org](https://www.python.org/).

#### **2. Установка pip**

**pip** — это пакетный менеджер для Python, который используется для установки библиотек и фреймворков, таких как Django. Обычно pip уже устанавливается вместе с Python, но если его нет, его можно установить вручную.

Чтобы проверить наличие pip, выполните в терминале команду:

```shell
pip --version
```

Если pip установлен, вы увидите номер его версии. Если же pip отсутствует, его можно установить с помощью команды:

```shell
python -m ensurepip --upgrade
```

#### **3. Установка virtualenv**

**virtualenv** — это инструмент, который позволяет создавать виртуальные окружения для Python-проектов. Виртуальные окружения помогают изолировать зависимости, что особенно важно, если у вас на компьютере одновременно разрабатываются несколько проектов с разными версиями библиотек и фреймворков. Виртуальные окружения предотвращают конфликты между проектами и позволяют легко управлять зависимостями.

Для установки virtualenv выполните команду:

```shell
pip install virtualenv
```

---

### **Почему мы используем виртуальные окружения?**

При разработке веб-приложений на Python, особенно с использованием Django, использование виртуальных окружений является критически важным аспектом. Виртуальные окружения создают изолированное пространство для каждого проекта, что предотвращает возникновение конфликтов между версиями зависимостей и упрощает управление проектами. Рассмотрим подробнее, почему виртуальные окружения так важны.

#### **1. Изоляция зависимостей**

Когда вы работаете над несколькими проектами одновременно, каждый из них может требовать разные версии библиотек или пакетов. Например, один проект может требовать Django версии 3.2, а другой — Django 4.0. Если не использовать виртуальные окружения, установка новой версии библиотеки может нарушить работу существующих проектов, так как они могут перестать работать с изменённой версией зависимостей.

Виртуальное окружение создаёт отдельную среду, в которой можно установить все зависимости для конкретного проекта. Это позволяет вам использовать разные версии одной и той же библиотеки в разных проектах без каких-либо конфликтов.

#### **2. Легкость в управлении зависимостями**

Использование виртуального окружения упрощает управление зависимостями проекта. Все установленные библиотеки сохраняются в отдельной папке, и вам не нужно беспокоиться о том, что случайное обновление библиотеки может повлиять на другие проекты.

Кроме того, виртуальные окружения делают процесс развёртывания проекта более предсказуемым, так как вы точно знаете, какие версии библиотек установлены и какие зависимости нужны для корректной работы приложения.

#### **3. Упрощение развёртывания и совместной работы**

Когда проект готов к развёртыванию на сервере или передаче другому разработчику, наличие виртуального окружения позволяет легко передать список зависимостей с помощью файла `requirements.txt`. Этот файл содержит все библиотеки, которые были установлены в виртуальном окружении, а другой разработчик или сервер могут просто установить те же зависимости, запустив команду:

```shell
pip install -r requirements.txt
```

Это гарантирует, что проект будет работать одинаково на разных машинах, так как везде будут установлены одинаковые версии библиотек.

#### **4. Минимизация рисков при обновлениях**

Обновление библиотек всегда связано с риском, что новые версии могут содержать несовместимые изменения. Используя виртуальные окружения, вы можете изолировать изменения, протестировать обновления в рамках одного проекта и убедиться, что обновления не повлияли на работу приложения. Таким образом, вы минимизируете риски, связанные с обновлением пакетов.

### **2.2 Установка Django**

После того как мы подготовили наше окружение (установили Python, pip и настроили виртуальное окружение), следующим шагом будет установка самого Django. Django устанавливается с помощью пакетного менеджера pip, что делает этот процесс простым и удобным. В этом разделе рассмотрим, как установить Django через pip, и почему важно следить за актуальностью версий.

#### **1. Установка Django через pip**

Для начала, убедитесь, что вы находитесь в созданном виртуальном окружении. Это можно сделать командой:

```shell
source env/bin/activate   # на MacOS/Linux
env\Scripts\activate      # на Windows
```

Теперь, чтобы установить последнюю стабильную версию Django, достаточно выполнить следующую команду:

```shell
pip install django
```

После успешной установки можно проверить, что Django установлен, и узнать его текущую версию с помощью команды:

```shell
django-admin --version
```

Эта команда отобразит версию Django, которая была установлена в вашем виртуальном окружении.


#### **2. Зачем следить за последними версиями Django?**

Использование актуальных версий Django — это важная практика для любого разработчика. Рассмотрим несколько причин, почему стоит регулярно обновлять фреймворк до последней стабильной версии.

##### **2.1 Доступ к новым функциям**

Каждый новый релиз Django включает в себя улучшения, новые функции и инструменты, которые делают разработку проще, быстрее и эффективнее. Например, новые версии могут предлагать улучшенные инструменты для работы с асинхронным кодом, более простую настройку или новые шаблонные теги.

Используя последнюю версию, вы всегда будете иметь доступ к этим новым возможностям, что упростит разработку и сделает ваш код более современным.

##### **2.2 Повышение безопасности**

Безопасность — одна из ключевых причин, почему необходимо регулярно обновлять Django. Разработчики Django уделяют большое внимание безопасности фреймворка и регулярно выпускают обновления, закрывающие уязвимости.

Устаревшие версии Django могут иметь известные уязвимости, которые могут быть использованы злоумышленниками для атак на ваше приложение. Использование последней версии помогает защитить ваше приложение от таких угроз. Например, Django активно защищает от распространённых уязвимостей, таких как SQL-инъекции, межсайтовая подделка запросов (CSRF) и межсайтовый скриптинг (XSS), но новые методы атаки могут появляться, и они исправляются в обновлениях фреймворка.

##### **2.3 Улучшение производительности**

Каждая новая версия Django включает оптимизации производительности, которые могут сделать ваши приложения быстрее и эффективнее. Это особенно важно для больших проектов, где производительность критична. Оптимизация запросов к базе данных, улучшенная работа с асинхронностью и кеширование — это только некоторые примеры улучшений, которые регулярно добавляются в новые версии.

##### **2.4 Поддержка со стороны сообщества**

Django — это фреймворк с активным и большим сообществом разработчиков. Это сообщество постоянно работает над улучшением фреймворка, а также над созданием новых плагинов и библиотек. Поддержка последних версий означает, что вы сможете легче находить помощь и документацию по новым возможностям, а также использовать сторонние библиотеки, которые также регулярно обновляются и адаптируются под новые версии Django.

##### **2.5 Устранение ошибок**

Даже в самом стабильном коде могут быть ошибки, и разработчики Django постоянно исправляют обнаруженные баги. Использование последней версии Django помогает избежать известных ошибок, которые могли быть найдены и устранены в предыдущих версиях.


#### **3. Как обновить Django?**

Если у вас уже установлена более старая версия Django, её можно легко обновить до последней версии с помощью команды:

```shell
pip install --upgrade django
```

Эта команда автоматически установит последнюю версию фреймворка. Чтобы убедиться, что обновление прошло успешно, можно снова проверить версию Django через команду:

```shell
django-admin --version
```


### **2.3 Создание и настройка проекта**

Теперь, когда мы установили Django, давайте перейдем к созданию первого проекта. В этой части мы рассмотрим, как сгенерировать базовую структуру проекта с помощью встроенной команды Django, разберём, какие файлы и папки создаются, и запустим сервер разработки, чтобы увидеть первую страницу приложения.

---

#### **1. Создание проекта с помощью команды `django-admin startproject`**

Django предлагает мощный инструмент командной строки — `django-admin`, с помощью которого можно создавать и управлять проектами. Для создания нового проекта используйте команду:

```shell
django-admin startproject myproject
```

Здесь `myproject` — это имя вашего проекта, которое вы можете заменить на любое другое. Эта команда создаст базовую структуру файлов и директорий, которые понадобятся для работы с Django.

##### **Что происходит при выполнении команды?**

Когда вы запускаете `django-admin startproject`, Django создаёт начальный набор файлов и директорий, которые представляют собой каркас вашего проекта. Эта структура уже настроена для работы и включает все необходимые компоненты для того, чтобы запустить проект.

---

#### **2. Обзор структуры проекта**

После создания проекта команда создаст следующие файлы и директории:

```shell
myproject/
    manage.py
    myproject/
        __init__.py
        settings.py
        urls.py
        asgi.py
        wsgi.py
```

Теперь давайте разберём, что делает каждый файл и папка:

##### **`manage.py`**
- Это ключевой скрипт для управления вашим проектом. Он используется для выполнения команд, таких как запуск сервера разработки, миграции базы данных и создания новых приложений. Команды `manage.py` тесно связаны с настройками проекта и позволяют легко взаимодействовать с Django.

##### **`myproject/`**
- Это главная директория проекта, которая содержит все основные конфигурационные файлы. Внутри неё находятся настройки проекта, маршрутизация URL и модули, необходимые для работы вашего приложения.

##### **`__init__.py`**
- Пустой файл, который отмечает директорию как Python-пакет. Это стандартный файл для любой папки Python, который позволяет импортировать её как модуль.

##### **`settings.py`**
- Один из самых важных файлов в проекте. Здесь находятся все настройки вашего проекта: подключение к базе данных, конфигурация приложений, настройки безопасности, языковые и временные зоны и многое другое. Этот файл будет регулярно изменяться по мере того, как ваш проект будет развиваться.

##### **`urls.py`**
- Этот файл отвечает за маршрутизацию в вашем приложении. В `urls.py` настраиваются пути (URL), которые будут обрабатываться различными представлениями (views) вашего проекта. По сути, здесь вы указываете, какая часть кода должна выполняться при переходе пользователя по определённому адресу.

##### **`wsgi.py`**
- Этот файл используется для развёртывания проекта на серверах, поддерживающих WSGI (Web Server Gateway Interface). WSGI — это стандарт для связи между веб-сервером и веб-приложением на Python. Этот файл вам понадобится, когда вы будете готовить проект для развёртывания в продакшн.

##### **`asgi.py`**
- ASGI (Asynchronous Server Gateway Interface) — это аналог WSGI, но для асинхронных приложений. Этот файл нужен для поддержки асинхронных запросов, что становится всё более важным в современных приложениях.

---

#### **3. Запуск сервера разработки и просмотр первой страницы проекта**

Django поставляется с встроенным сервером разработки, который позволяет вам тестировать ваше приложение без необходимости настраивать полноценный веб-сервер. Это делает процесс разработки быстрым и удобным.

Чтобы запустить сервер разработки, выполните команду:

```shell
python manage.py runserver
```

После запуска сервера вы увидите что-то похожее на следующее сообщение в терминале:

```shell
Watching for file changes with StatReloader
Performing system checks...

System check identified no issues (0 silenced).
September 8, 2024 - 12:34:56
Django version 4.2.3, using settings 'myproject.settings'
Starting development server at http://127.0.0.1:8000/
Quit the server with CONTROL-C.
```

Сервер разработки Django работает на порту `8000` и доступен по адресу `http://127.0.0.1:8000/`. Если вы откроете этот адрес в браузере, вы увидите первую страницу Django, которая подтвердит успешную установку и запуск проекта.

##### **Что мы видим на первой странице?**

Открыв в браузере `http://127.0.0.1:8000/`, вы увидите стандартную стартовую страницу Django с заголовком "The install worked successfully! Congratulations!". Это подтверждение того, что ваше приложение настроено правильно, и сервер работает.
