## **5. Работа с моделями и БД**

Работа с моделями и базами данных — одна из ключевых тем в разработке на Django, особенно в контексте создания интернет-магазина. Модели представляют собой способ работы с данными, абстрагируя нас от необходимости напрямую писать SQL-запросы. Django позволяет легко создавать структуры данных в виде классов Python и синхронизировать их с базой данных, используя миграции.

---

### **5.1 Что такое модель в Django?**

Модели в Django — это классы, которые описывают структуру данных. Каждая модель соответствует таблице в базе данных, а атрибуты класса модели — это столбцы этой таблицы. Модели помогают описать, как данные будут храниться, извлекаться и изменяться.

- **Пример модели**:
  Рассмотрим пример модели для интернет-магазина, которая описывает товары:
  ```python
  from django.db import models

  class Product(models.Model):
      name = models.CharField(max_length=255)
      description = models.TextField()
      price = models.DecimalField(max_digits=10, decimal_places=2)
      stock = models.PositiveIntegerField()
      available = models.BooleanField(default=True)
      created_at = models.DateTimeField(auto_now_add=True)
      updated_at = models.DateTimeField(auto_now=True)

      def __str__(self):
          return self.name
  ```

  **Основные поля модели**:
  - `name`: Название товара (строка с максимальной длиной 255 символов).
  - `description`: Описание товара (поле для текста).
  - `price`: Цена товара с точностью до двух знаков после запятой.
  - `stock`: Количество товара на складе.
  - `available`: Статус доступности товара (True или False).
  - `created_at` и `updated_at`: Автоматически сохраняемые временные метки создания и обновления записи.

  Эта модель представляет структуру данных для хранения товаров, которая затем будет отображаться в базе данных как таблица с соответствующими полями.

---

### **5.2 Миграции в Django**

Миграции — это механизм Django для управления изменениями в базе данных. Когда вы создаете или изменяете модели, Django автоматически генерирует миграции, которые преобразуют эти изменения в SQL-запросы для базы данных.

#### **Основные этапы работы с миграциями**:

1. **Создание миграции**: Когда вы изменяете модели, вы должны создать миграцию, которая описывает эти изменения. Команда для создания миграции:
   ```bash
   python manage.py makemigrations
   ```
   Эта команда сканирует изменения в ваших моделях и создает файлы миграций, которые будут применяться к базе данных.

2. **Применение миграции**: После создания миграции необходимо применить её к базе данных, чтобы изменения вступили в силу:
   ```bash
   python manage.py migrate
   ```
   Эта команда выполняет все миграции, которые еще не были применены, синхронизируя структуру базы данных с моделями.

#### **Типы миграций**:

- **Начальная миграция**: Создается, когда модель впервые добавляется в проект. Она создаёт таблицу в базе данных.
- **Миграции изменений**: Создаются при изменении существующих моделей (например, добавление нового поля или изменение типа данных).
- **Миграции удаления**: Удаляют поля или модели из базы данных.

### **5.3 Пример создания модели**

Рассмотрим пример создания модели для интернет-магазина, которая будет хранить данные о категориях товаров. Мы также создадим миграции и применим их, чтобы соответствующая таблица появилась в базе данных.

##### **Шаг 1: Создание модели**

Создадим модель `Category` в файле `models.py`, которая будет описывать категории товаров. Например:

```python
from django.db import models

class Category(models.Model):
    name = models.CharField(max_length=100)
    slug = models.SlugField(unique=True)
    description = models.TextField(blank=True, null=True)
    created_at = models.DateTimeField(auto_now_add=True)

    def __str__(self):
        return self.name
```

##### **Пояснение полей**:
- **name**: Название категории (строковое поле с ограничением в 100 символов).
- **slug**: Человекочитаемый идентификатор, который будет использоваться в URL (например, `electronics`).
- **description**: Дополнительное описание категории (опционально, может быть пустым).
- **created_at**: Дата и время создания категории (автоматически заполняется при создании записи).

##### **Шаг 2: Создание миграции**

После добавления модели нужно создать миграцию:

```bash
python manage.py makemigrations
```

Эта команда создаст файл миграции, который будет содержать SQL-запрос для создания таблицы `Category` в базе данных.

##### **Шаг 3: Применение миграции**

Теперь необходимо применить миграцию, чтобы таблица категории появилась в базе данных:

```bash
python manage.py migrate
```

После выполнения этой команды в базе данных будет создана таблица `category` с полями, которые мы определили в модели.

##### **Шаг 4: Использование модели**

Теперь мы можем работать с моделью через ORM Django. Например, создать новую категорию:

```python
from catalog.models import Category

new_category = Category(name="Электроника", slug="electronics")
new_category.save()  # Сохраняет запись в базе данных
```

##### **Шаг 5: Использование в админке**

Чтобы управлять категориями через административную панель Django, зарегистрируем модель в файле `admin.py`:

```python
from django.contrib import admin
from .models import Category

@admin.register(Category)
class CategoryAdmin(admin.ModelAdmin):
    list_display = ('name', 'slug', 'created_at')
    search_fields = ('name', 'slug')
```

Теперь в панели администратора можно будет добавлять, редактировать и удалять категории товаров.
