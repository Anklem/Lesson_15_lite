В Django HTML-шаблоны используют специальный синтаксис для встраивания кода Python и взаимодействия с данными. Этот синтаксис, называемый Django Template Language (DTL), позволяет динамически генерировать HTML-страницы на основе данных, которые передаются из представлений (views).

### Основные элементы синтаксиса DTL

#### 1. **Вывод переменных**

Чтобы отобразить переменную, которая была передана в шаблон из представления, используется двойное фигурное заключение:

```html
{{ variable_name }}
```

Например, если в представление передана переменная `post.title`, то в шаблоне можно вывести её так:

```html
<h1>{{ post.title }}</h1>
```

#### 2. **Фильтры для переменных**

Фильтры позволяют модифицировать выводимые данные. Они применяются через символ `|`. Вот несколько примеров фильтров:

- **`|lower`** — приводит строку к нижнему регистру:
  ```html
  <p>{{ post.title|lower }}</p>
  ```

- **`|date`** — форматирует дату:
  ```html
  <p>Опубликовано: {{ post.created_at|date:"d M Y" }}</p>
  ```

- **`|length`** — возвращает длину строки или списка:
  ```html
  <p>Количество символов в заголовке: {{ post.title|length }}</p>
  ```

#### 3. **Циклы `for`**

Циклы позволяют итерировать по спискам или другим итераторам. Синтаксис цикла в шаблонах Django следующий:

```html
<ul>
  {% for post in posts %}
    <li>{{ post.title }}</li>
  {% endfor %}
</ul>
```

Если список пуст, можно добавить блок `{% empty %}`:

```html
<ul>
  {% for post in posts %}
    <li>{{ post.title }}</li>
  {% empty %}
    <li>Нет доступных постов</li>
  {% endfor %}
</ul>
```

#### 4. **Условные операторы `if`**

Условные операторы позволяют проверять значения и выводить данные в зависимости от условия:

```html
{% if post.is_published %}
  <p>Эта статья опубликована</p>
{% else %}
  <p>Эта статья пока не опубликована</p>
{% endif %}
```

Можно использовать более сложные условия:

```html
{% if post.is_published and post.is_featured %}
  <p>Эта статья опубликована и находится в избранном</p>
{% endif %}
```

#### 5. **Наследование шаблонов**

Django поддерживает концепцию "наследования шаблонов", которая позволяет создавать базовые шаблоны и наследовать их в других шаблонах. Базовый шаблон содержит общий каркас страницы (например, шапку и подвал), а дочерние шаблоны могут переопределять определенные блоки.

##### Пример базового шаблона `base.html`:

```html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>{% block title %}My Blog{% endblock %}</title>
</head>
<body>
    <header>
        <h1>Мой блог</h1>
    </header>

    <main>
        {% block content %}{% endblock %}
    </main>

    <footer>
        <p>© 2024 My Blog</p>
    </footer>
</body>
</html>
```

##### Пример шаблона, который наследует `base.html`:

```html
{% extends 'base.html' %}

{% block title %}Главная страница{% endblock %}

{% block content %}
    <h2>Список постов</h2>
    <ul>
        {% for post in posts %}
            <li>{{ post.title }}</li>
        {% endfor %}
    </ul>
{% endblock %}
```

#### 6. **Подключение статических файлов**

Для подключения CSS, JavaScript, изображений и других статических ресурсов используется тег `{% static %}`. Пример подключения CSS:

```html
<link rel="stylesheet" href="{% static 'css/styles.css' %}">
```

Для работы с `{% static %}` нужно убедиться, что в `settings.py` настроена директория для статических файлов, и добавлена строка `django.contrib.staticfiles` в `INSTALLED_APPS`.

Пример конфигурации в `settings.py`:

```python
STATIC_URL = '/static/'
```

А также нужно подключить `{% load static %}` в шаблоне:

```html
{% load static %}
```

#### 7. **Подключение URL-ов в шаблонах**

Для генерации ссылок на страницы используется тег `{% url %}`. Он позволяет создавать ссылки на определенные маршруты по их именам:

```html
<a href="{% url 'home' %}">На главную</a>
```

Здесь `'home'` — это имя маршрута, заданное в файле `urls.py`.

### Пример полного шаблона

Пример простого HTML-шаблона с использованием DTL для отображения списка постов:

```html
{% extends 'base.html' %}

{% block title %}Главная страница{% endblock %}

{% block content %}
    <h1>Список постов</h1>
    <ul>
        {% for post in posts %}
            <li>
                <h2>{{ post.title }}</h2>
                <p>Опубликовано: {{ post.created_at|date:"d M Y" }}</p>
                <p>{{ post.content }}</p>
            </li>
        {% empty %}
            <p>Пока нет постов.</p>
        {% endfor %}
    </ul>
{% endblock %}
```

Этот шаблон выводит список постов блога, отображает дату публикации в формате `день месяц год`, и если нет постов, выводит сообщение "Пока нет постов".
