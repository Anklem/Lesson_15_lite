## **4. Создание первого приложения**

Создание приложения в Django — это важный этап разработки, особенно если вы создаёте интернет-магазин. Приложения в Django — это отдельные модули, которые легко масштабировать и развивать. В контексте интернет-магазина, приложения могут быть предназначены для работы с товарами, корзиной, оформлением заказов, пользователями и т.д. Рассмотрим подробно, как создавать и структурировать приложения для подобных целей.

---

### **4.1 Что такое приложения в Django?**

Приложения в Django представляют собой независимые компоненты проекта, которые могут быть использованы для определенной функциональности. Проект может содержать несколько приложений, каждое из которых отвечает за конкретную задачу. Например, в интернет-магазине вы можете создать следующие приложения:

- **catalog** — для управления товарами.
- **cart** — для управления корзиной покупок.
- **checkout** — для оформления и обработки заказов.
- **user** — для управления пользователями и аутентификацией.

##### **Ключевые преимущества использования приложений**:
- **Модульность**: каждое приложение легко расширять и тестировать отдельно.
- **Переиспользуемость**: приложения могут быть использованы в других проектах.
- **Удобство разработки**: структура проекта становится более понятной и управляемой.

---

### **4.2 Структура приложения**

Каждое приложение в Django имеет стандартную структуру файлов и папок, которые упрощают разработку и организацию кода. Рассмотрим основные компоненты структуры приложения на примере создания приложения для управления товарами в интернет-магазине.

1. **models.py** — содержит описание моделей, которые представляют собой структуры данных для базы данных. Для интернет-магазина здесь будут модели для хранения информации о товарах, категориях и, возможно, отзывы.

   - **Пример модели товара**:
     ```python
     from django.db import models

     class Product(models.Model):
         name = models.CharField(max_length=255)
         description = models.TextField()
         price = models.DecimalField(max_digits=10, decimal_places=2)
         stock = models.PositiveIntegerField()
         created_at = models.DateTimeField(auto_now_add=True)
         updated_at = models.DateTimeField(auto_now=True)
     ```

     Эта модель представляет таблицу в базе данных для хранения информации о товаре, включая название, описание, цену и количество на складе.

2. **views.py** — отвечает за логику обработки запросов. Здесь мы будем определять представления, которые взаимодействуют с пользователем, обрабатывают запросы и возвращают соответствующие ответы. Например, представление для отображения списка товаров или для отображения страницы с деталями конкретного товара.

   - **Пример представления для отображения списка товаров**:
     ```python
     from django.shortcuts import render
     from .models import Product

     def product_list(request):
         products = Product.objects.all()
         return render(request, 'catalog/product_list.html', {'products': products})
     ```

3. **urls.py** — файл для маршрутизации запросов внутри приложения. Здесь мы определяем, какие URL должны обрабатывать какие представления. Например, URL для отображения списка товаров или детальной страницы товара.

   - **Пример маршрутизации**:
     ```python
     from django.urls import path
     from . import views

     urlpatterns = [
         path('products/', views.product_list, name='product_list'),
         path('products/<int:pk>/', views.product_detail, name='product_detail'),
     ]
     ```

     Здесь мы создаем два маршрута: один для списка всех товаров, а другой — для страницы с деталями товара по его идентификатору.

4. **admin.py** — файл, где мы регистрируем модели для отображения их в административной панели Django. Это удобно для управления товарами и их атрибутами через интерфейс администратора.

   - **Пример регистрации модели в админке**:
     ```python
     from django.contrib import admin
     from .models import Product

     @admin.register(Product)
     class ProductAdmin(admin.ModelAdmin):
         list_display = ('name', 'price', 'stock', 'created_at')
     ```

     Это позволит вам видеть товары в панели администратора, управлять их атрибутами и отслеживать их состояние (например, цену и количество на складе).

---

### **4.3 Подключение приложения к проекту**

Чтобы новое приложение стало частью вашего проекта, его необходимо зарегистрировать и подключить к общему маршрутизатору URL. Рассмотрим этот процесс пошагово.

##### **1. Регистрация приложения в проекте**

Для того чтобы приложение стало активным в проекте, его нужно добавить в список установленных приложений (INSTALLED_APPS) в файле `settings.py`.

- **Добавление приложения в `INSTALLED_APPS`**:
  ```python
  INSTALLED_APPS = [
      'django.contrib.admin',
      'django.contrib.auth',
      'django.contrib.contenttypes',
      'django.contrib.sessions',
      'django.contrib.messages',
      'django.contrib.staticfiles',
      'catalog',  # Добавляем наше приложение "catalog"
  ]
  ```

##### **2. Подключение маршрутов приложения к проекту**

Маршруты для каждого приложения могут быть подключены к общему файлу `urls.py` на уровне проекта. Это позволит интегрировать URL вашего приложения в общую систему маршрутизации.

- **Пример подключения маршрутов приложения к проекту**:
  В файле `urls.py` на уровне проекта добавьте следующее:
  ```python
  from django.contrib import admin
  from django.urls import path, include

  urlpatterns = [
      path('admin/', admin.site.urls),
      path('catalog/', include('catalog.urls')),  # Подключаем маршруты приложения "catalog"
  ]
  ```

Теперь все запросы, начинающиеся с `/catalog/`, будут обрабатываться в приложении `catalog`.

### **4.4 Создание первого представления**

Теперь, когда приложение подключено, давайте создадим наше первое представление. Представления — это функции или классы, которые обрабатывают HTTP-запросы и возвращают HTTP-ответы. Мы начнем с простого представления, которое вернёт текст на веб-страницу.

##### **1. Создание простого представления**

Создадим простое представление, которое вернёт строку «Добро пожаловать в интернет-магазин!» на экран.

- **Пример представления в `views.py`**:
  ```python
  from django.http import HttpResponse

  def welcome_view(request):
      return HttpResponse("Добро пожаловать в интернет-магазин!")
  ```

##### **2. Подключение представления к маршруту**

Добавим маршрут, который будет обрабатывать запросы к этому представлению. Для этого изменим файл `urls.py` в приложении:

- **Пример маршрута в `urls.py`**:
  ```python
  from django.urls import path
  from . import views

  urlpatterns = [
      path('welcome/', views.welcome_view, name='welcome'),
  ]
  ```

Теперь при переходе по URL `http://127.0.0.1:8000/catalog/welcome/` на экране появится сообщение «Добро пожаловать в интернет-магазин!».

##### **3. Работа с шаблонами**

Django поддерживает систему шаблонов, которая позволяет рендерить HTML-страницы. Вместо простого текстового ответа можно использовать шаблоны для отображения данных и структуры страницы.

- **Пример представления с использованием шаблона**:
  ```python
  from django.shortcuts import render

  def welcome_view(request):
      return render(request, 'catalog/welcome.html')
  ```

- **Пример шаблона (welcome.html)**:
  ```html
  <!DOCTYPE html>
  <html>
  <head>
      <title>Добро пожаловать</title>
  </head>
  <body>
      <h1>Добро пожаловать в интернет-магазин!</h1>
  </body>
  </html>
  ```

Этот подход позволяет вам легко управлять дизайном и содержимым страниц.
