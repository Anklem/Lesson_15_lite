## **3. Создание первого проекта**


Теперь, когда мы установили Django и создали новый проект, важно разобраться в структуре проекта, чтобы понять, как он работает. В этой секции подробно рассмотрим ключевые файлы и директории, которые создаются при запуске команды `django-admin startproject`, и объясним их назначение.

---

### **3.1 Структура проекта Django**

После создания проекта с помощью `django-admin startproject <project_name>`, структура проекта будет выглядеть примерно так:

```
myproject/
    manage.py
    myproject/
        __init__.py
        settings.py
        urls.py
        asgi.py
        wsgi.py
```

Разберем каждую часть этой структуры:

#### **1. `manage.py`**

- **Описание**: Это основной скрипт для управления вашим проектом Django. Он обеспечивает доступ ко многим командам Django, таким как запуск сервера разработки, создание миграций, запуск тестов и другие задачи.
- **Назначение**: Этот файл используется для выполнения различных административных задач в проекте. Например, вы можете использовать его для создания суперпользователя, выполнения миграций и запуска сервера разработки.
- **Пример использования**:
  - Запуск сервера разработки:
    ```
    python manage.py runserver
    ```
  - Создание миграций:
    ```
    python manage.py makemigrations
    ```

#### **2. `myproject/` (директория проекта)**

Эта директория содержит основные конфигурационные файлы вашего проекта и имеет ту же самую структуру, что и имя проекта. Все важные настройки и маршрутизация находятся здесь.

##### **2.1 `__init__.py`**

- **Описание**: Пустой файл, который указывает Python, что эта директория является пакетом.
- **Назначение**: Этот файл не содержит кода и служит только для того, чтобы Python распознавал эту директорию как модуль. Это позволяет вам импортировать другие модули и пакеты из этой директории.

##### **2.2 `settings.py`**

- **Описание**: Это один из самых важных файлов в вашем проекте. Он содержит все настройки вашего проекта, такие как конфигурация базы данных, настройки безопасности, подключение приложений и другие параметры.
- **Назначение**: `settings.py` определяет, как проект будет взаимодействовать с внешними системами и как будет вести себя в разных условиях.
- **Ключевые секции**:
  - **DATABASES**: Конфигурация для подключения к базе данных. Например:
    ```python
    DATABASES = {
        'default': {
            'ENGINE': 'django.db.backends.sqlite3',
            'NAME': BASE_DIR / 'db.sqlite3',
        }
    }
    ```
  - **INSTALLED_APPS**: Список приложений, которые активны в проекте. Например:
    ```python
    INSTALLED_APPS = [
        'django.contrib.admin',
        'django.contrib.auth',
        'django.contrib.contenttypes',
        'django.contrib.sessions',
        'django.contrib.messages',
        'django.contrib.staticfiles',
    ]
    ```
  - **STATIC_URL**: Определяет URL для статических файлов. Например:
    ```python
    STATIC_URL = '/static/'
    ```
  - **SECRET_KEY**: Секретный ключ, используемый для криптографических операций.

##### **2.3 `urls.py`**

- **Описание**: Этот файл содержит маршрутизацию URL для вашего проекта. Здесь определяется, какие запросы должны обрабатываться какими представлениями.
- **Назначение**: `urls.py` устанавливает, какой код будет выполняться при запросах к определённым URL-адресам. Это позволяет связывать URL-адреса с функциями и классами представлений.
- **Пример использования**:
  ```python
  from django.contrib import admin
  from django.urls import path

  urlpatterns = [
      path('admin/', admin.site.urls),
  ]
  ```
  Здесь `admin/` будет обрабатываться административным сайтом Django.

##### **2.4 `wsgi.py`**

- **Описание**: Этот файл используется для запуска вашего проекта на сервере, который поддерживает WSGI (Web Server Gateway Interface). WSGI — это стандартный интерфейс для связи между веб-сервером и веб-приложением на Python.
- **Назначение**: `wsgi.py` предоставляет точку входа для WSGI-сервера, чтобы он мог передать запросы вашему приложению. Это важно при развертывании проекта на сервере, таком как Gunicorn или uWSGI.
- **Пример использования**:
  ```python
  import os
  from django.core.wsgi import get_wsgi_application

  os.environ.setdefault('DJANGO_SETTINGS_MODULE', 'myproject.settings')
  application = get_wsgi_application()
  ```

##### **2.5 `asgi.py`**

- **Описание**: Этот файл используется для запуска вашего проекта на сервере, поддерживающем ASGI (Asynchronous Server Gateway Interface). ASGI является стандартом для асинхронных веб-приложений на Python и поддерживает асинхронные запросы и WebSocket.
- **Назначение**: `asgi.py` предоставляет точку входа для ASGI-сервера, позволяя обрабатывать асинхронные запросы, что полезно для приложений, требующих высокой производительности.
- **Пример использования**:
  ```python
  import os
  from django.core.asgi import get_asgi_application

  os.environ.setdefault('DJANGO_SETTINGS_MODULE', 'myproject.settings')
  application = get_asgi_application()
  ```

---

### **Вывод:**

Разобравшись в структуре проекта Django, вы теперь имеете четкое представление о роли каждого ключевого файла и директории. `manage.py` управляет административными задачами, `settings.py` содержит настройки, `urls.py` управляет маршрутизацией, а `wsgi.py` и `asgi.py` обеспечивают поддержку различных серверов для развертывания. Понимание этой структуры поможет вам эффективно разрабатывать и поддерживать ваше Django-приложение.

### **3.2 Как работает Django?**

Чтобы понять, как работает Django, важно рассмотреть основные этапы обработки запросов и взаимодействия с базой данных. Мы рассмотрим, как Django обрабатывает HTTP-запросы от клиента, как система маршрутизации направляет запросы к соответствующим представлениям и как ORM (Object-Relational Mapping) упрощает взаимодействие с базами данных.

---

#### **1. Обработка запросов в Django**

Давайте пройдемся по этапам обработки запроса, чтобы увидеть, как Django управляет запросами и возвращает ответы.

##### **1.1. Клиентский запрос**

Когда клиент (например, пользовательский браузер) отправляет HTTP-запрос к вашему Django-приложению, запрос сначала попадает на сервер, на котором развернут Django-проект. 

##### **1.2. URL-диспетчер**

Сервер передает запрос в Django, который начинается с обработки маршрутизации URL. Здесь происходит следующее:

- **URLconf**: Django использует файл `urls.py` в проекте для определения, какому представлению (view) должен быть передан запрос. Этот файл содержит список маршрутов (URL patterns), которые сопоставляют URL-запросы с функциями или классами представлений.

- **Пример URLconf**:
  ```python
  from django.contrib import admin
  from django.urls import path
  from myapp import views

  urlpatterns = [
      path('admin/', admin.site.urls),
      path('hello/', views.hello_world, name='hello_world'),
  ]
  ```

  В этом примере запрос к `http://127.0.0.1:8000/hello/` будет передан функции `hello_world` в `views.py` вашего приложения.

##### **1.3. Представление (View)**

После того как URL-адрес сопоставлен с представлением, Django вызывает соответствующую функцию или метод класса. Представление обрабатывает запрос и подготавливает ответ. В зависимости от логики приложения, представление может делать следующее:

- Получать данные из базы данных.
- Выполнять бизнес-логику.
- Генерировать HTML-шаблон или другой формат ответа (например, JSON).

- **Пример представления**:
  ```python
  from django.http import HttpResponse

  def hello_world(request):
      return HttpResponse("Hello, world!")
  ```

  Это простое представление возвращает текстовый ответ "Hello, world!".

##### **1.4. Шаблоны и ответы**

Если представление требует рендеринга HTML-страницы, оно передает данные в шаблон. Django использует систему шаблонов для генерации HTML-ответов. Шаблоны содержат HTML-код с динамическими частями, которые заменяются данными из представления.

- **Пример шаблона (hello.html)**:
  ```html
  <!DOCTYPE html>
  <html>
  <head>
      <title>Hello</title>
  </head>
  <body>
      <h1>Hello, {{ name }}!</h1>
  </body>
  </html>
  ```

- **Пример представления, использующего шаблон**:
  ```python
  from django.shortcuts import render

  def hello_world(request):
      context = {'name': 'world'}
      return render(request, 'hello.html', context)
  ```

  Здесь `render` передает контекст в шаблон `hello.html`, который заменяет `{{ name }}` на `world`.

##### **1.5. Возврат ответа**

После того как представление подготовило ответ, он возвращается в Django, который затем отправляет его обратно клиенту. В случае HTML-шаблонов это будет готовый HTML-код, а в случае API — JSON-объект.

---

#### **2. Object-Relational Mapping (ORM)**

Django включает мощный инструмент ORM, который упрощает взаимодействие с базами данных, позволяя работать с данными как с объектами, а не через прямые SQL-запросы. 

##### **2.1. Модели**

В Django модели представляют собой классы, которые определяют структуру таблиц в базе данных. Каждая модель соответствует одной таблице, а атрибуты модели представляют собой столбцы в таблице.

- **Пример модели**:
  ```python
  from django.db import models

  class Person(models.Model):
      name = models.CharField(max_length=100)
      age = models.IntegerField()
  ```

  Эта модель создаст таблицу `person` в базе данных с двумя столбцами: `name` и `age`.

##### **2.2. Миграции**

После определения модели вам нужно создать и применить миграции, чтобы синхронизировать вашу базу данных с определением модели.

- **Создание миграций**:
  ```python
  python manage.py makemigrations
  ```

  Эта команда создаёт файлы миграций на основе изменений в моделях.

- **Применение миграций**:
  ```python
  python manage.py migrate
  ```

  Эта команда применяет миграции и создаёт или изменяет таблицы в базе данных.

##### **2.3. Запросы через ORM**

Django ORM позволяет выполнять запросы к базе данных с помощью Python-кода, а не SQL-запросов. Например:

- **Создание объекта**:
  ```python
  person = Person(name="Alice", age=30)
  person.save()
  ```

- **Получение объекта**:
  ```python
  person = Person.objects.get(id=1)
  ```

- **Фильтрация объектов**:
  ```python
  young_people = Person.objects.filter(age__lt=30)
  ```

- **Обновление объекта**:
  ```python
  person.age = 31
  person.save()
  ```

- **Удаление объекта**:
  ```python
  person.delete()
  ```

##### **2.4. Запросы с использованием Django Admin**

Django предоставляет встроенный интерфейс администрирования, который позволяет управлять данными через веб-интерфейс. Для его использования необходимо зарегистрировать модели в `admin.py`.

- **Пример регистрации модели**:
  ```python
  from django.contrib import admin
  from .models import Person

  @admin.register(Person)
  class PersonAdmin(admin.ModelAdmin):
      list_display = ('name', 'age')
  ```

  После регистрации модели вы сможете управлять данными через административный интерфейс по адресу `http://127.0.0.1:8000/admin/`.

---

### **Вывод:**

Django обрабатывает запросы от клиента через систему маршрутизации URL, передавая их соответствующим представлениям, которые формируют ответ. ORM упрощает взаимодействие с базами данных, позволяя работать с данными через модели и Python-код вместо прямых SQL-запросов. Это делает разработку более удобной и эффективной, избавляя вас от необходимости писать сложные SQL-запросы вручную.


### **3.3 Настройки проекта**

Настройка проекта Django включает в себя конфигурацию баз данных, установку приложений и настройку параметров, необходимых для успешной работы вашего приложения. В этой секции мы подробно рассмотрим, как настроить базу данных, используя стандартные и альтернативные варианты, а также как работать с встроенными приложениями Django.

---

#### **1. Настройка базы данных**

По умолчанию Django использует SQLite в качестве базы данных. Это простое решение подходит для разработки и тестирования. Однако для продакшн-окружения часто выбирают более масштабируемые решения, такие как PostgreSQL или MySQL.

##### **1.1. Использование SQLite**

SQLite — это встроенная в Python база данных, которая хранит данные в одном файле на диске. Чтобы использовать SQLite, не нужно ничего настраивать, так как Django уже настроен для работы с ней по умолчанию.

- **Конфигурация** в `settings.py`:
  ```python
  DATABASES = {
      'default': {
          'ENGINE': 'django.db.backends.sqlite3',
          'NAME': BASE_DIR / 'db.sqlite3',
      }
  }
  ```

  Здесь `ENGINE` указывает, что используется SQLite, а `NAME` задаёт имя файла базы данных.

##### **1.2. Использование PostgreSQL**

PostgreSQL — мощная реляционная база данных, которая хорошо поддерживает масштабируемость и сложные запросы. Чтобы использовать PostgreSQL, нужно установить дополнительный пакет и настроить `settings.py`.

- **Установка зависимостей**:
  ```
  pip install psycopg2-binary
  ```

- **Конфигурация** в `settings.py`:
  ```python
  DATABASES = {
      'default': {
          'ENGINE': 'django.db.backends.postgresql',
          'NAME': 'mydatabase',
          'USER': 'myuser',
          'PASSWORD': 'mypassword',
          'HOST': 'localhost',
          'PORT': '5432',
      }
  }
  ```

  Здесь `ENGINE` указывает на PostgreSQL, а остальные параметры (`NAME`, `USER`, `PASSWORD`, `HOST`, `PORT`) задают данные для подключения к базе данных.

##### **1.3. Использование MySQL**

MySQL — ещё одна популярная реляционная база данных. Для её использования потребуется дополнительный пакет и соответствующая настройка.

- **Установка зависимостей**:
  ```
  pip install mysqlclient
  ```

- **Конфигурация** в `settings.py`:
  ```python
  DATABASES = {
      'default': {
          'ENGINE': 'django.db.backends.mysql',
          'NAME': 'mydatabase',
          'USER': 'myuser',
          'PASSWORD': 'mypassword',
          'HOST': 'localhost',
          'PORT': '3306',
      }
  }
  ```

  Здесь параметры аналогичны тем, что мы указывали для PostgreSQL.

---

#### **2. Встроенные приложения Django**

Django предоставляет несколько встроенных приложений, которые облегчают разработку и помогают ускорить создание веб-приложений. Рассмотрим некоторые из них:

##### **2.1. Административный интерфейс**

Django включает в себя административный интерфейс, который позволяет управлять моделями вашего приложения через удобный веб-интерфейс.

- **Настройка**: Админка включена по умолчанию в проекте. Чтобы использовать её, необходимо зарегистрировать модели в файле `admin.py` вашего приложения.
  
  - **Пример регистрации модели**:
    ```python
    from django.contrib import admin
    from .models import Person

    @admin.register(Person)
    class PersonAdmin(admin.ModelAdmin):
        list_display = ('name', 'age')
    ```

  После регистрации вы сможете получить доступ к админке по адресу `http://127.0.0.1:8000/admin/`.

##### **2.2. Аутентификация**

Django предоставляет встроенные средства для управления пользователями и их аутентификацией. Это включает регистрацию, вход, выход, изменение пароля и управление пользователями.

- **Настройка**: Аутентификация включена по умолчанию. Основные компоненты включают:
  - `django.contrib.auth`: Модели для пользователей и групп, а также представления и формы для регистрации и входа.
  - `django.contrib.sessions`: Управление сессиями пользователей.
  - `django.contrib.messages`: Система сообщений для отображения уведомлений пользователям.

- **Пример использования**: Вы можете использовать встроенные представления для аутентификации пользователей:
  ```python
  from django.contrib.auth import views as auth_views

  urlpatterns = [
      path('login/', auth_views.LoginView.as_view(), name='login'),
      path('logout/', auth_views.LogoutView.as_view(), name='logout'),
  ]
  ```

##### **2.3. Статические файлы и медиа**

Django управляет статическими файлами (CSS, JavaScript) и медиафайлами (загрузками пользователей) с помощью встроенных настроек.

- **Статические файлы**: Настройки для статических файлов находятся в `settings.py`:
  ```python
  STATIC_URL = '/static/'
  ```

  Статические файлы собираются в одну директорию при помощи команды `collectstatic`:
  ```
  python manage.py collectstatic
  ```

- **Медиафайлы**: Настройки для медиафайлов включают путь для хранения загруженных файлов:
  ```python
  MEDIA_URL = '/media/'
  MEDIA_ROOT = BASE_DIR / 'media'
  ```

  Эти файлы будут сохраняться в директории `media` в корне вашего проекта.

##### **2.4. Миддлвары**

Миддлвары — это компоненты, которые обрабатывают запросы и ответы в процессе их прохождения через приложение. Django включает несколько встроенных миддлваров, которые обеспечивают базовую функциональность, такую как обработка сессий и кук.

- **Пример миддлвара**: `django.middleware.security.SecurityMiddleware` обеспечивает базовые меры безопасности, такие как добавление заголовков безопасности к ответам.

- **Конфигурация** миддлваров находится в `settings.py` в разделе `MIDDLEWARE`:
  ```python
  MIDDLEWARE = [
      'django.middleware.security.SecurityMiddleware',
      'django.contrib.sessions.middleware.SessionMiddleware',
      'django.middleware.common.CommonMiddleware',
      'django.middleware.csrf.CsrfViewMiddleware',
      'django.contrib.auth.middleware.AuthenticationMiddleware',
      'django.contrib.messages.middleware.MessageMiddleware',
      'django.middleware.clickjacking.XFrameOptionsMiddleware',
  ]
  ```

### **Вывод:**

Настройки проекта Django позволяют адаптировать ваше приложение под разные базы данных и использовать встроенные инструменты для упрощения разработки. Конфигурация базы данных может варьироваться в зависимости от ваших требований, будь то SQLite, PostgreSQL или MySQL. Встроенные приложения, такие как административный интерфейс и аутентификация, предоставляют удобные и мощные средства для управления проектом, делая разработку более быстрой и эффективной.
