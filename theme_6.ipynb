## **6. Django Admin и работа с данными**

Административная панель Django — это встроенный инструмент для управления данными через удобный графический интерфейс. В отличие от большинства других фреймворков, Django сразу предоставляет готовое решение для управления моделями и данными, что значительно ускоряет разработку.

---

### **6.1 Введение в Django Admin**

Админка Django — это мощный инструмент для управления данными проекта без необходимости писать код. Это веб-интерфейс, который позволяет создавать, редактировать, удалять и просматривать записи баз данных, а также управлять пользователями и правами доступа.

**Основные возможности админки Django**:
- **Просмотр данных**: Вы можете легко просматривать записи, сортировать и фильтровать их.
- **Редактирование данных**: Записи могут быть отредактированы или удалены прямо через админку.
- **Управление пользователями и правами**: Администраторы могут управлять доступом к различным частям сайта и назначать роли.
- **Модернизация**: Админка поддерживает кастомизацию — можно изменять внешний вид и функциональность для управления различными моделями.

**Использование админки на примере интернет-магазина**:
- Управление товарами, категориями, пользователями и заказами происходит через админку, что делает её отличным инструментом для небольших и средних проектов.
- Можно легко добавить новые товары, редактировать описания или изменять статус заказов.

Админка автоматически генерируется на основе моделей, которые вы создаёте в проекте. Она позволяет разработчикам сосредоточиться на функциональной части приложения, предоставляя готовый интерфейс для работы с данными.

---

### **6.2 Создание суперпользователя**

Чтобы получить доступ к административной панели Django, необходимо создать суперпользователя — пользователя с полными правами для управления всеми аспектами проекта через админку.

**Процесс создания суперпользователя**:

1. Откройте терминал и выполните команду:

   ```bash
   python manage.py createsuperuser
   ```

2. Django запросит несколько полей, необходимых для создания суперпользователя:
   - **Username**: Имя пользователя (например, `admin`).
   - **Email address**: Электронная почта (например, `admin@example.com`).
   - **Password**: Пароль (с учетом стандартов безопасности).

   После успешного ввода данных суперпользователь будет создан и записан в базу данных.

3. Запустите сервер разработки:

   ```bash
   python manage.py runserver
   ```

4. Откройте административную панель в браузере по адресу: `http://127.0.0.1:8000/admin/`. Используйте данные суперпользователя для входа в систему.

Теперь вы сможете увидеть интерфейс админки и работать с моделями, которые зарегистрированы в панели администратора.

---

### **6.3 Регистрация моделей в Django Admin**

Чтобы модели, которые мы создаем в проекте, отображались в админке, их необходимо зарегистрировать в файле `admin.py` каждого приложения. Это дает возможность управлять данными через интерфейс, редактировать записи и настраивать отображение данных.

**Пример регистрации модели**:

Рассмотрим, как зарегистрировать модель `Product` (которая описывает товары для интернет-магазина) в админке.

1. Откройте файл `admin.py` в папке приложения (например, в приложении `catalog`):
   ```python
   from django.contrib import admin
   from .models import Product
   ```

2. Зарегистрируйте модель:
   ```python
   @admin.register(Product)
   class ProductAdmin(admin.ModelAdmin):
       list_display = ('name', 'price', 'available', 'stock')
       list_filter = ('available', 'created_at')
       search_fields = ('name', 'description')
   ```

**Пояснение параметров**:
- `list_display`: Определяет, какие поля модели будут отображаться в таблице на главной странице админки.
- `list_filter`: Добавляет фильтры по полям, чтобы администратор мог быстро отфильтровать товары по их доступности или дате создания.
- `search_fields`: Позволяет искать записи по указанным полям (например, по имени товара или описанию).

Теперь, когда модель зарегистрирована, администратор может:
- Просматривать список товаров с основными данными (название, цена, количество на складе).
- Фильтровать товары по их доступности и дате создания.
- Выполнять поиск по названию и описанию товаров.

**Расширенные возможности админки**:
- **Редактирование вложенных данных**: Если модель имеет связи с другими моделями (например, товар принадлежит категории), можно настроить админку для управления связанными объектами.
- **Подключение графиков и визуализаций**: Возможно добавить инструменты для отображения статистики по продажам или активности пользователей прямо в админку.

### **Пример расширенной кастомизации админки для интернет-магазина**

Для интернет-магазина можно настроить админку таким образом, чтобы администратор имел полный контроль над товарами, категориями и заказами. Например, для управления заказами можно добавить дополнительные действия (например, массовое изменение статуса заказов) или включить инлайновое редактирование связанных данных (товары в заказе).

Пример кастомизации для модели заказа:

```python
from django.contrib import admin
from .models import Order, OrderItem

class OrderItemInline(admin.TabularInline):
    model = OrderItem
    extra = 1  # Количество дополнительных пустых строк для новых товаров в заказе

@admin.register(Order)
class OrderAdmin(admin.ModelAdmin):
    list_display = ('id', 'customer', 'status', 'created_at')
    list_filter = ('status', 'created_at')
    search_fields = ('customer__username', 'id')
    inlines = [OrderItemInline]
```

**Описание**:
- **OrderItemInline**: Это инлайн-редактор, который позволяет добавлять или редактировать товары, связанные с конкретным заказом, прямо на странице редактирования заказа.
- **list_display**: Отображает ключевые поля заказа (номер заказа, клиент, статус).
- **list_filter**: Позволяет фильтровать заказы по статусу или дате создания.

Таким образом, админка для интернет-магазина становится не просто панелью управления данными, а полноценным инструментом для работы с заказами, пользователями и товарами, позволяя легко управлять всеми аспектами интернет-магазина через удобный интерфейс.
